## Process ActorAttack Results

```python
import json
import os
result_path = 'results/multi_turn/XX_model.json'

score = 0
with open(result_path) as f:
    data = json.load(f)
# 200 attack target goal, each has 3 attempts
for i in range(0,600,3):
    for j in range(3):
        if data[i+j]['score'] == 1:
            score += 1
            break
score /= 200
print(score)
```

## Process Over-Refusal Results

```python
import json
import os

json_files = ['xstest_0.json', 'OKTest_0.json', 'ORbenc_0.json', 'PHtest_0.json']  

folder_path='results/overrefusal/XX_model'
results = []
for file in json_files:
    file_path = os.path.join(folder_path, file)
    with open(file_path, 'r') as f:
        data = json.load(f)
        results.append(data['score'])
print(results)
```
